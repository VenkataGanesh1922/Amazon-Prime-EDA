```python
import os
import ast
import pandas as pd
import matplotlib.pyplot as plt

# Create graphs folder if it doesn't exist
os.makedirs("../graphs", exist_ok=True)

# Function to safely convert list-like dataset values
def parse_list(value):
    if isinstance(value, list):
        return value

    if pd.isna(value):
        return []

    try:
        result = ast.literal_eval(str(value))
        if isinstance(result, list):
            return result
        return []
    except (ValueError, SyntaxError):
        return []

# Reusable pie chart function
def create_pie_chart(counts, title, filename, top_n=8):

    counts = counts.dropna()
    counts = counts[counts > 0].sort_values(ascending=False)

    if counts.empty:
        print(f"No data available for {title}")
        return

    # Keep top categories and group the rest as Others
    if len(counts) > top_n:
        top_counts = counts.head(top_n).copy()
        others = counts.iloc[top_n:].sum()

        if others > 0:
            top_counts.loc["Others"] = others

        counts = top_counts

    plt.figure(figsize=(9, 8))

    plt.pie(
        counts,
        labels=counts.index,
        autopct="%1.1f%%",
        startangle=90,
        pctdistance=0.75
    )

    plt.title(title, fontsize=15, fontweight="bold")
    plt.tight_layout()

    plt.savefig(
        f"../graphs/{filename}",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    print("\n", title)
    print(counts)
```

```python

# CHART 14: CORRELATION HEATMAP

import matplotlib.pyplot as plt
import seaborn as sns
import os

# Select numerical columns
numeric_data = titles_clean[
    ['release_year', 'runtime', 'imdb_score',
     'imdb_votes', 'tmdb_popularity', 'tmdb_score']
].copy()

# Convert columns into numeric format
numeric_data = numeric_data.apply(
    pd.to_numeric, errors='coerce'
)

# Calculate correlation
correlation = numeric_data.corr()

# Display correlation values
print("CORRELATION MATRIX:")
display(correlation.round(2))

# Create heatmap
plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    square=True
)

plt.title(
    'Correlation Heatmap of Amazon Prime Movies and TV Shows',
    fontsize=14,
    fontweight='bold'
)

plt.tight_layout()

# Save chart
os.makedirs("../graphs", exist_ok=True)

plt.savefig(
    "../graphs/14_correlation_heatmap.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()

print("Chart 14 saved successfully!")
```

```python

# CHART 15: PAIR PLOT ANALYSIS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Select important numerical variables
pair_data = titles_clean[
    [
        'release_year',
        'runtime',
        'imdb_score',
        'tmdb_score',
        'tmdb_popularity'
    ]
].copy()

# Convert values into numeric format
pair_data = pair_data.apply(
    pd.to_numeric, errors='coerce'
)

# Remove missing values
pair_data = pair_data.dropna()

# Take a sample for faster visualization
pair_data = pair_data.sample(
    n=min(1000, len(pair_data)),
    random_state=42
)

# Generate pair plot
pair_plot = sns.pairplot(
    pair_data,
    diag_kind='hist',
    corner=True,
    plot_kws={'alpha': 0.5, 's': 15}
)

# Add overall title
pair_plot.fig.suptitle(
    'Pair Plot Analysis of Amazon Prime Movies and TV Shows',
    fontsize=16,
    fontweight='bold',
    y=1.02
)

# Save chart
os.makedirs("../graphs", exist_ok=True)

pair_plot.fig.savefig(
    "../graphs/15_pair_plot.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()

print("Chart 15 saved successfully!")
```
